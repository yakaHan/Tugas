```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/uiuc-cse/data-fa14/gh-pages/data/iris.csv")
```

```python
print(df.info())
```

```python
# a) Label Encoding untuk kolom species
le = LabelEncoder()
df['species_encoded'] = le.fit_transform(df['species'])

# b) Scaling dengan StandardScaler
scaler = StandardScaler()
df[['sepal_length_scaled', 'petal_width_scaled']] = scaler.fit_transform(df[['sepal_length', 'petal_width']])

# c) Menampilkan 5 baris pertama hasil preprocessing
print("5 baris pertama hasil preprocessing:")
print(df.head())
```

```python
print(df[['species', 'species_encoded', 'sepal_length', 'sepal_length_scaled',
          'petal_width', 'petal_width_scaled']].head())
```

```python
dt = pd.read_csv("/content/Book1.csv", delimiter=';')
print(dt)
```

```python
print(dt.info())
```

```python
dt = pd.DataFrame(dt)
```

```python
#a) One-Hot Encoding untuk kolom "City"
dt_encoded = pd.get_dummies(dt, columns=['City '])

# b) Normalisasi harga menggunakan MinMaxScaler
scaler = MinMaxScaler()
dt_encoded['Price_normalized'] = scaler.fit_transform(dt_encoded[['Price']])

# c) Menampilkan dataset akhir
print("Dataset akhir setelah preprocessing:")
print(dt_encoded)
```

Label Encoding dan One-Hot Encoding adalah dua metode untuk mengubah data kategorikal menjadi numerik, tetapi dengan pendekatan berbeda. Label Encoding mengkonversi setiap kategori menjadi angka unik (misal: Jakarta=0, Surabaya=1), cocok untuk variabel target atau data ordinal yang memiliki hierarki, tetapi berisiko menimbulkan misinterpretasi model karena angka dianggap memiliki urutan. One-Hot Encoding membuat kolom terpisah untuk setiap kategori dengan nilai 0 atau 1 (misal: kolom "Jakarta" bernilai 1 jika benar), ideal untuk data nominal tanpa urutan seperti nama kota, karena menghindari asumsi ranking, tetapi dapat meningkatkan dimensi data secara signifikan jika kategori banyak. Label Encoding lebih efisien untuk kategori yang sangat banyak atau variabel target, sementara One-Hot Encoding lebih aman untuk model yang sensitif terhadap urutan numerik seperti regresi linear atau KNN, selama jumlah kategori tidak terlalu besar (kurang dari 15). Pemilihan metode tergantung pada jenis data, jumlah kategori, dan algoritma yang digunakan.
