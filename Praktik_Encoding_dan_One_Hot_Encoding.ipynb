```
# Label Encoding
le = LabelEncoder()
df['type_encoded'] = le.fit_transform(df['type'])
print("=== Label Encoding ===")
print(df[['type', 'type_encoded']].head())

# One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=['rating'], drop_first=True)
print("\n=== One-Hot Encoding ===")
print(df_encoded.filter(like='rating_').head())

# Scaling
scaler = MinMaxScaler()
df['release_year_normalized'] = scaler.fit_transform(df[['release_year']])

scaler = StandardScaler()
df['release_year_standardized'] = scaler.fit_transform(df[['release_year']])

print("\n=== Scaling Results ===")
print(df[['release_year', 'release_year_normalized', 'release_year_standardized']].head())
```
